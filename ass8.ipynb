```python
import seaborn as sns
import matplotlib.pyplot as plt
from seaborn import set_style
from matplotlib import rcParams
import pandas as pd
```

```python
# Load the Titanic dataset from Seaborn
titanic_data = sns.load_dataset('titanic')
```

```python
# Remove rows with any missing values
titanic_data_cleaned = titanic_data.dropna()
```

```python
titanic_data_cleaned.isnull().sum()
```

```python
# Set the style of the plot
set_style('whitegrid')

# Define plot parameters
rcParams['figure.figsize'] = 10, 6

# Plot histogram of ticket prices
sns.histplot(titanic_data['fare'], bins=20, kde=True)
plt.title('Distribution of Ticket Prices')
plt.xlabel('Fare')
plt.ylabel('Frequency')
plt.show()
```

```python
# Joint plot between age and fare
sns.jointplot(x='age', y='fare', data=titanic_data_cleaned, kind='scatter')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.title('Joint Plot of Age and Fare')
plt.show()
```

```python
# Pairplot between age, fare, pclass, and survived
sns.pairplot(titanic_data[['age', 'fare', 'pclass', 'survived']], hue='survived')
plt.suptitle('Pairplot of Age, Fare, Pclass, and Survived')
plt.show()
```

```python
# Barplot of passenger class
sns.barplot(x='age', y='sex', data=titanic_data)
plt.title('Barplot between age and sex')
plt.xlabel('Age')
plt.ylabel('Sex')
plt.show()

```

```python
# Countplot of embarked locations
sns.countplot(x='sex', data=titanic_data)
plt.title('Count of Passengers by Sex')
plt.xlabel('Sex')
plt.ylabel('Count')
plt.show()
```
