```python
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from sklearn.feature_extraction.text import TfidfVectorizer
import string
```

```python
# Read text from file
file_path = "7_story.txt"
with open(file_path, 'r') as file:
    text = file.read()
```

```python
# Step 3: Perform Tokenization
# A. Sentence Tokenization
sentences = sent_tokenize(text)
print("Sentence Tokenization:")
print(sentences)
```

```python
# B. Word Tokenization
tokens = word_tokenize(text)
print("\nWord Tokenization:")
print(tokens)
```

```python
# Step 4 A: Removing Punctuations
# Remove punctuation
table = str.maketrans('', '', string.punctuation)
text_no_punct = text.translate(table)
```

```python
# Step 4 B: Stop Words
# Get English stop words
stop_words = set(stopwords.words('english'))

# Remove stop words
filtered_tokens = [word for word in word_tokenize(text_no_punct) if word.lower() not in stop_words]
print("\nFiltered Tokens (without Punctuation and Stop Words):")
print(filtered_tokens)
```

```python
# Step 5: Perform Lemmatization
# Initialize WordNet Lemmatizer
lemmatizer = WordNetLemmatizer()

# Perform Lemmatization
lemmatized_tokens = [lemmatizer.lemmatize(word) for word in filtered_tokens]
print("\nLemmatized Tokens:")
print(lemmatized_tokens)
```

```python
# Step 6: Apply POS Tagging
pos_tags = nltk.pos_tag(filtered_tokens)
print("\nPOS Tagging:")
print(pos_tags)
```

```python
# Step 7: Calculate Term Frequency and Inverse Document Frequency (TF-IDF)
# Convert list of tokens to a string for TF-IDF calculation
text_for_tfidf = ' '.join(lemmatized_tokens)
```

```python
# Initialize TF-IDF vectorizer
tfidf_vectorizer = TfidfVectorizer()
```

```python
# Fit and transform the text
tfidf_matrix = tfidf_vectorizer.fit_transform([text_for_tfidf])
```

```python
# Get TF-IDF representation
tfidf_representation = tfidf_matrix.toarray()
```

```python
print("\nTF-IDF Representation:")
print(tfidf_representation)
```
