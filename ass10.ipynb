```python
# 2. Create a histogram for each feature
iris_df.hist(figsize=(10, 8))
plt.suptitle("Histogram for each feature in the Iris dataset", y=0.95)
plt.show()
```

```python
# Select columns to include in the box plot (excluding 'id' column)
columns_to_plot = iris_df.columns[iris_df.columns != 'Id']

# Create a box plot for selected features
plt.figure(figsize=(10, 8))
sns.boxplot(data=iris_df[columns_to_plot])
plt.title("Box plot for each feature in the Iris dataset")
plt.xticks(rotation=45)
plt.show()
```
